```python

import pandas as bbi

df = bbi.read_excel ('tarefa02.xlsx')
display (df.head(21))
```

```python
#Gráfico Pizza/Gráfico de Seções

import matplotlib.pyplot as plt

course_counts = df['CURSO'].value_counts()

plt.figure(figsize=(10, 8))
plt.pie(course_counts, labels=course_counts.index, autopct='%1.1f%%', startangle=90, textprops={'fontsize':10})
plt.title('Distribuição de Cursos', fontsize= 16)
plt.axis ('equal')
plt.show()
```

```python
# Boxplot

import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 8))
sns.boxplot(data=df, x= 'CURSO', y= 'IDADE', palette= 'pastel')
plt.title('Distribuição de Idade por Curso', fontsize=16)
plt.xlabel ('Curso', fontsize=12)
plt.ylabel ('Idade', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
#Gráfico de Barras

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_excel('tarefa02.xlsx') # Ensure df is defined

plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='CURSO', palette='viridis', order = df['CURSO'].value_counts().index)
plt.title('Distribuição de Alunos por Curso', fontsize=16)
plt.xlabel('Curso', fontsize=12)
plt.ylabel('Número de Alunos', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
#Histograma

import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 8))
sns.histplot(data=df, x= 'IDADE', bins=range(df['IDADE'].min(), df['IDADE'].max() + 2), kde=True)
plt.title("Histograma de Distribuição de Idade", fontsize=16)
plt.xlabel('Idade', fontsize=12)
plt.ylabel('Frequência de Aparição', fontsize=12)
plt.xticks(range(df['IDADE'].min(), df['IDADE'].max() + 1))
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```
