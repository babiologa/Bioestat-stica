```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
df = pd.read_excel("tarefa03.xlsx")
display(df)
```

```python
freq_tipotosse = df["TIPO DE TOSSE"].value_counts()
freq_tipotosse
```

```python
freq_tosse = df["VOCÊ ESTÁ COM TOSSE?"].value_counts()
freq_tosse
```

```python
plt.figure(figsize=(8, 6))
sns.barplot(x=freq_tipotosse.index, y=freq_tipotosse.values, palette='viridis')
plt.title('Frequência do Tipo de Tosse')
plt.xlabel('Tipo de Tosse')
plt.ylabel('Número de Alunos')
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.histplot(
    data=df,
    x="TIPO DE TOSSE",
    hue="VOCÊ ESTÁ COM TOSSE?",
    multiple="stack",
    kde=False
)
plt.title("Relação da frequência da tosse com o tipo de tosse")
plt.xlabel("Tipo")
plt.ylabel("Frequência")
plt.tight_layout()
plt.show()
```

```python
grouped_counts = df.groupby(['VOCÊ ESTÁ COM TOSSE?', 'TIPO DE TOSSE']).size().reset_index(name='Numero de Alunos')

plt.figure(figsize=(8, 6))
sns.scatterplot(
    data=grouped_counts,
    x="TIPO DE TOSSE",
    y="Numero de Alunos",
    hue="VOCÊ ESTÁ COM TOSSE?",
)
plt.title("Relação entre Resposta à Tosse, Tipo de Tosse e Número de Alunos")
plt.xlabel("Tipo de Tosse")
plt.ylabel("Número de Alunos")
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(title='Você está com Tosse?', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
