#4 -
#a - A distribuição caso o pesquisador analise apenas uma árvore é a de Bernoulli, pois a árvore pode estar somente infestada ou não infestada.

#b - A distribuição é a Binomial, pois o pesquisador quer saber quantas vezes há sucesso dentro de um evento (a árvore estar infestada), em um espaço amostral.

#c - Na distribuição de Bernoulli, há apenas um experimento feito, com dois resultados possíveis. Já na distribuição Binomial, há mais de um experimento feito, e o que queremos analisar é a ocorrência de um só resultado, em todos os experimentos, diferentemente da distribuição de Bernoulli, que pretende examinar qual é o resultado do experimento dentro de dois resultados possíveis.

```python
from scipy.stats import bernoulli, binom, poisson

#Questão 2: Binomial n=10, p=0.3
n, p = 10, 0.3
print("2b  P(X=3)  =", round(binom.pmf(3, n, p), 4))                # ~0.2668
print("2c  P(X>=1) =", round(1 - binom.pmf(0, n, p), 4))            # ~0.9718
# alternativa: binom.sf(0, n, p)

# Questão 3: Poisson lambda=2
lam = 2
print("3b  P(X=3)  =", round(poisson.pmf(3, lam), 4))               # ~0.1804
print("3c  P(X=0)  =", round(poisson.pmf(0, lam), 4))               # ~0.1353
print("3d  P(X>2)  =", round(poisson.sf(2, lam), 4))                # ~0.3233
# sf(k) = 1 - cdf(k) = P(X > k)

#Questão 4 (exemplo com p = 0,2)
p_inf = 0.2
print("4a  Bernoulli P(infestada) =", bernoulli.pmf(1, p_inf))
print("4b  Binomial(50, p): média =", binom.mean(50, p_inf),
      "| variância =", binom.var(50, p_inf))
```
