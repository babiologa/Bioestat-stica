## Exercício 1

### a) Hipóteses

- **H₀:** μ = 2,0 cm
- **H₁:** μ ≠ 2,0 cm

O teste é bilateral porque a pergunta é se a média é diferente de 2,0 cm, sem especificar se seria maior ou menor.

### b) Teste utilizado

Usamos o **teste t de Student para uma amostra**, porque queremos comparar a média da amostra com um valor de referência, o desvio-padrão populacional é desconhecido e a amostra é pequena. O enunciado também informa que a população tem distribuição aproximadamente normal.

```python
import numpy as np
from scipy import stats

crescimento = np.array([1.8, 2.1, 2.4, 1.9, 2.2, 2.0, 2.5, 1.7, 2.3, 2.1])

n = len(crescimento)
media = np.mean(crescimento)
desvio = np.std(crescimento, ddof=1)

print("Número de mudas:", n)
print("Média amostral:", round(media, 3), "cm")
print("Desvio-padrão amostral:", round(desvio, 3), "cm")
```

### c) Média e desvio-padrão amostral
A média é calculada somando os valores e dividindo pelo número de mudas. Para o desvio-padrão amostral, usamos `ddof=1` no Python.

### d) Estatística t
A fórmula é \(t = (\bar{x} - \mu_0)/(s/\sqrt{n})\).

```python
media_referencia = 2.0
t_calculado = (media - media_referencia) / (desvio / np.sqrt(n))

print("Estatística t:", round(t_calculado, 3))
```

### e) Valor crítico e f) decisão
Com nível de significância de 5% e 9 graus de liberdade, o teste é bilateral. Rejeitamos H₀ se o valor de t ficar abaixo de −t crítico ou acima de +t crítico.

```python
import numpy as np
from scipy import stats

try:
    t_calculado
except NameError:
    media_referencia = 2.0
    t_calculado = (media - media_referencia) / (desvio / np.sqrt(n))

alpha = 0.05
gl = n - 1
t_critico = stats.t.ppf(1 - alpha/2, gl)
p_valor_t = 2 * stats.t.sf(abs(t_calculado), gl)

print("Graus de liberdade:", gl)
print("Valores críticos: ±", round(t_critico, 3))
print("p-valor:", round(p_valor_t, 4))

if abs(t_calculado) > t_critico:
    print("Decisão: rejeitar H0.")
else:
    print("Decisão: não rejeitar H0.")
```

### g) Conclusão

A média observada foi de aproximadamente **2,10 cm**, mas a estatística t calculada ficou dentro da região de não rejeição. Assim, com 5% de significância, não há evidência suficiente para afirmar que o crescimento médio mensal das mudas seja diferente de 2,0 cm.

## Exercício 2

### a) Hipóteses e b) tipo de teste

- **H₀:** p = 0,70
- **H₁:** p > 0,70

O teste é **unilateral à direita**, pois a afirmação que queremos verificar é se a proporção de germinação é superior a 70%.

```python
n_sementes = 100
germinadas = 78
p0 = 0.70
alpha = 0.05

p_amostral = germinadas / n_sementes
z_calculado = (p_amostral - p0) / np.sqrt(p0 * (1 - p0) / n_sementes)
z_critico = stats.norm.ppf(1 - alpha)
p_valor_z = stats.norm.sf(z_calculado)

print("Proporção amostral:", round(p_amostral, 3))
print("Estatística z:", round(z_calculado, 3))
print("Valor crítico:", round(z_critico, 3))
print("p-valor:", round(p_valor_z, 4))
```

### c) a f) Proporção, estatística, valor crítico e decisão

A proporção amostral é \(\hat p = 78/100 = 0,78\), ou seja, 78%. Pela aproximação normal, calculamos a estatística z e comparamos com o valor crítico de um teste unilateral à direita, com 5% de significância.

```python
if z_calculado > z_critico:
    print("Decisão: rejeitar H0.")
else:
    print("Decisão: não rejeitar H0.")
```

### g) Conclusão

Como a estatística z calculada é maior que o valor crítico, rejeitamos H₀. Portanto, ao nível de 5% de significância, os resultados fornecem evidências suficientes para apoiar a afirmação de que a proporção de germinação é superior a 70%.

## Exercício 3

### a) Hipóteses

- **H₀:** σ² = 0,25 (mg/L)²
- **H₁:** σ² > 0,25 (mg/L)²

O teste é unilateral à direita porque queremos verificar se a variância é maior que o valor de referência.

### b) Teste utilizado

Usamos o **teste qui-quadrado para uma variância**, pois a hipótese trata diretamente da variância populacional. O teste t é usado principalmente para testar médias, e não é o mais adequado para esta pergunta. O teste qui-quadrado pressupõe que os dados tenham distribuição aproximadamente normal.

```python
n_pontos = 12
variancia_amostral = 0.40
variancia_referencia = 0.25
alpha = 0.05

gl_chi = n_pontos - 1
chi_calculado = gl_chi * variancia_amostral / variancia_referencia
chi_critico = stats.chi2.ppf(1 - alpha, gl_chi)
p_valor_chi = stats.chi2.sf(chi_calculado, gl_chi)

print("Estatística qui-quadrado:", round(chi_calculado, 3))
print("Graus de liberdade:", gl_chi)
print("Valor crítico:", round(chi_critico, 3))
print("p-valor:", round(p_valor_chi, 4))
```

### c) a f) Estatística, graus de liberdade, valor crítico e decisão

A estatística é calculada por \(\chi^2 = (n-1)s^2/\sigma_0^2\). Como são 12 pontos, temos 11 graus de liberdade. No teste unilateral à direita, rejeitamos H₀ se a estatística calculada for maior que o valor crítico.

```python
if chi_calculado > chi_critico:
    print("Decisão: rejeitar H0.")
else:
    print("Decisão: não rejeitar H0.")
```

### g) Conclusão

A estatística qui-quadrado calculada é menor que o valor crítico. Por isso, não rejeitamos H₀. Com os dados disponíveis e ao nível de 5% de significância, não há evidências suficientes para afirmar que a variância da concentração de nitrato no riacho seja maior que 0,25 (mg/L)².
